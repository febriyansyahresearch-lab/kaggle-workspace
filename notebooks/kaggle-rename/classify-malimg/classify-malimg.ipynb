# Langkah 1 — Tampilkan gambar

```python
import os

image_files = []

for root, dirs, files in os.walk("/kaggle/input"):
    for f in files:
        if f.lower().endswith((".png", ".jpg", ".jpeg", ".bmp")):
            image_files.append(os.path.join(root, f))

print("Jumlah gambar ditemukan:", len(image_files))

for i in range(min(10, len(image_files))):
    print(image_files[i])
```

# Langkah 2 — Tampilkan label malware family

```python
from PIL import Image
import matplotlib.pyplot as plt
import os

img_path = image_files[0]
family_label = os.path.basename(os.path.dirname(img_path))

img = Image.open(img_path)

plt.figure(figsize=(5,5))
plt.imshow(img, cmap="gray")
plt.title(f"Malware Family: {family_label}")
plt.axis("off")
plt.show()

print("Path gambar    :", img_path)
print("Family label   :", family_label)
```

# Langkah 3 — Tunjukkan metadata dataset

```python
import os
import pandas as pd

base_dir = "/kaggle/input"

family_counts = []

for root, dirs, files in os.walk(base_dir):
    image_files = [
        f for f in files
        if f.lower().endswith((".png", ".jpg", ".jpeg", ".bmp"))
    ]

    if len(image_files) > 0:
        family_name = os.path.basename(root)

        family_counts.append({
            "Family": family_name,
            "Images": len(image_files)
        })

df = pd.DataFrame(family_counts)

df = df.sort_values(
    by="Images",
    ascending=False
).reset_index(drop=True)

print("Jumlah Malware Families :", len(df))
print("Total Malware Images    :", df["Images"].sum())

display(df)
```

```python
import os

families = {}

for root, dirs, files in os.walk("/kaggle/input"):
    imgs = [
        f for f in files
        if f.lower().endswith((".png",".jpg",".jpeg",".bmp"))
    ]

    if len(imgs) > 0:
        family = os.path.basename(root)
        families[family] = len(imgs)

print("Jumlah Family Malware :", len(families))
print("Total Images          :", sum(families.values()))

print("\n10 Family Pertama:\n")

for k,v in list(families.items())[:10]:
    print(f"{k:20} {v}")
```

# Langkah 3 — Visualisasi distribusi family malware

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

df.plot(
    x="Family",
    y="Images",
    kind="bar",
    legend=False
)

plt.title("Distribusi Malware Family pada Dataset MalImg")
plt.ylabel("Jumlah Images")
plt.tight_layout()

plt.show()
```
